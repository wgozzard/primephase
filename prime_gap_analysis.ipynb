# PrimePhase: Prime Gap Analysis Notebook

**This notebook accompanies the PrimePhase paper and performs:**
- Prime number generation
- Gap calculation
- Segment-wise average and standard deviation
- Chi-squared testing vs linear growth
- FFT analysis to detect harmonic structure

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from scipy.stats import chisquare
import pandas as pd
```

## Generate First 10,000 Primes

```python
primes = list(sp.primerange(0, 105000))
primes = primes[:10000]
gaps = np.diff(primes)
```

## Segment Gaps by 50-Prime Blocks and Analyze

```python
block_size = 50
avg_gaps, std_gaps, labels = [], [], []
ranges = [(i, i + block_size) for i in range(0, len(gaps), block_size)]
for start, end in ranges:
    local_gaps = gaps[start:end]
    if len(local_gaps) == 0:
        continue
    avg_gaps.append(np.mean(local_gaps))
    std_gaps.append(np.std(local_gaps))
    labels.append(f"{start}-{end}")
```

## Chi-Squared Test vs Linear Growth Model

```python
expected = np.linspace(avg_gaps[0], avg_gaps[-1], len(avg_gaps))
scaled_expected = expected * (np.sum(avg_gaps) / np.sum(expected))
chi_stat, p_val = chisquare(f_obs=avg_gaps, f_exp=scaled_expected)
chi_stat, p_val
```

## FFT Analysis of Prime Gaps

```python
fft = np.fft.fft(avg_gaps)
freq = np.fft.fftfreq(len(avg_gaps), d=1)
plt.figure(figsize=(12,6))
plt.plot(freq[:len(freq)//2], np.abs(fft)[:len(freq)//2])
plt.title("FFT of Average Prime Gaps (50-Block Segments)")
plt.xlabel("Frequency")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```
